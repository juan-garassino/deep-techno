# Data Exploration — MAESTRO + all_notes.csv

Run `00_setup_and_data.ipynb` first to set `ALL_NOTES`, `MIDI_DIR`, and install the package.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# ── Adjust paths if needed ───────────────────────────────────────
MIDI_DIR  = '/content/maestro'
ALL_NOTES = '/content/drive/MyDrive/deep-techno-data/all_notes.csv'  # or /content/all_notes.csv
MAESTRO_META = os.path.join(MIDI_DIR, 'maestro-v2.0.0', 'maestro-v2.0.0.csv')
```

## MAESTRO Metadata

```python
meta = pd.read_csv(MAESTRO_META)
print(f"Rows: {len(meta)}, Columns: {list(meta.columns)}")
meta.head()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 4))

# Composer distribution
meta['canonical_composer'].value_counts().head(20).plot(kind='barh', ax=axes[0])
axes[0].set_title('Top 20 composers'); axes[0].set_xlabel('# pieces')

# Duration distribution
axes[1].hist(meta['duration'] / 60, bins=40, color='steelblue', edgecolor='white')
axes[1].set_title('Piece duration distribution'); axes[1].set_xlabel('minutes')

plt.tight_layout(); plt.show()
print(f"Total audio: {meta['duration'].sum()/3600:.1f} hours")
```

## all_notes.csv

```python
df = pd.read_csv(ALL_NOTES)
print(f"Shape: {df.shape}")
df.describe()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

df['pitch'].plot(kind='hist', bins=88, ax=axes[0], color='coral')
axes[0].set_title('Pitch distribution'); axes[0].set_xlabel('MIDI pitch')

df['step'].clip(0, 2).plot(kind='hist', bins=80, ax=axes[1], color='mediumseagreen')
axes[1].set_title('Step distribution (clipped 0-2s)'); axes[1].set_xlabel('seconds')

df['duration'].clip(0, 3).plot(kind='hist', bins=80, ax=axes[2], color='mediumpurple')
axes[2].set_title('Duration distribution (clipped 0-3s)'); axes[2].set_xlabel('seconds')

plt.tight_layout(); plt.show()
```

## Piano roll — first 200 notes

```python
sample = df.head(200)
plt.figure(figsize=(20, 4))
for _, row in sample.iterrows():
    plt.barh(row['pitch'], row['duration'], left=row['start'], height=0.8,
             color='steelblue', alpha=0.7)
plt.xlabel('Time (s)'); plt.ylabel('MIDI pitch')
plt.title('Piano roll — first 200 notes')
plt.tight_layout(); plt.show()
```
